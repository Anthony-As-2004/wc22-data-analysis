```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('wc22_match_data.csv')
```

```python
df[['home_goals', 'away_goals']] = df['score'].str.extract(r'(\d+)[^\d]+(\d+)').astype(int)
```

```python
df_home = df[['match', 'home_team', 'home_possession', 'home_goals', 'away_goals']].copy()
df_home.rename(columns={
    'home_team': 'team',
    'home_possession': 'possession',
    'home_goals': 'goals_for',
    'away_goals': 'goals_against'
}, inplace=True)
df_home['mando'] = 'Home'
```

```python
df_away = df[['match', 'away_team', 'away_possession', 'away_goals', 'home_goals']].copy()
df_away.rename(columns={
    'away_team': 'team',
    'away_possession': 'possession',
    'away_goals': 'goals_for',
    'home_goals': 'goals_against'
}, inplace=True)
df_away['mando'] = 'Away'
```

```python
df_final = pd.concat([df_home, df_away], ignore_index=True)
```

```python
df_final['resultado'] = np.where(df_final['goals_for'] > df_final['goals_against'], 'Vitória',
                        np.where(df_final['goals_for'] < df_final['goals_against'], 'Derrota', 'Empate'))
```

```python
df_final.to_csv('wc22_dados_limpos.csv', index=False)
```
